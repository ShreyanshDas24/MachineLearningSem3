```python
class CustomKMeans:
  
  def __init__(self,cluster_count,max_iterations):
    self.cluster_count=cluster_count
    self.max_iterations=max_iterations
    self.cluster_centers=None

  def fit_and_predict(self,feature_matrix):
    random_indices=np.random.choice(range(0,feature_matrix.shape[0]),self.cluster_count)
    self.cluster_centers=feature_matrix.iloc[random_indices].values
    for feature_idx in range(self.max_iterations):
      cluster_assignments=self.assign_cluster_labels(feature_matrix)
      previous_centers=self.cluster_centers
      self.cluster_centers=self.update_centers(feature_matrix,cluster_assignments)
      if np.all(previous_centers==self.cluster_centers):
        break
    return cluster_assignments
  
  def assign_cluster_labels(self,feature_matrix):
    cluster_assignments=[]
    distance_values=[]
    for data_row in feature_matrix.values:
      for center in self.cluster_centers:
        distance_values.append(np.sqrt(np.dot(data_row-center,data_row-center)))
      closest_distance=min(distance_values)
      closest_index=distance_values.index(closest_distance)
      cluster_assignments.append(closest_index)
      distance_values.clear()
    return np.array(cluster_assignments)
  
  def update_centers(self,feature_matrix,cluster_assignments):
    updated_centers= []
    cluster_labels=np.unique(cluster_assignments)
    for label in cluster_labels:
      updated_center=np.mean(feature_matrix.iloc[cluster_assignments==label].values, axis=0)
      updated_centers.append(updated_center)
    return np.array(updated_centers)
```

```python
kmeans_model=CustomKMeans(3,100)
cluster_results=kmeans_model.fit_and_predict(feature_matrix)
```

```python
print(cluster_results)
```

```python
feature_matrix['clusters']=cluster_results
```
