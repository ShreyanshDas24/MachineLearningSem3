```python
import numpy as np
import pandas as pd
import random
import matplotlib.pyplot as plt
```

```python
class PageRank:
  def __init__(self, network, iterations):
    self.network = network
    self.vertices = list(network.keys())
    self.vertex_count = len(self.vertices)
    self.iterations = iterations
    self.scores = {vertex: 1 / self.vertex_count for vertex in self.vertices}
    self.damping = 0.85

  def page_ranking(self):
    for _ in range(self.iterations):
      updated_scores = {}
      for vertex in self.vertices:
        scores = (1 - self.damping) / self.vertex_count
        for source_vertex in self.vertices:
          if vertex in self.network[source_vertex]:
            scores += self.damping * self.scores[source_vertex] / len(self.network[source_vertex])
        updated_scores[vertex] = scores
      self.scores = updated_scores
    return self.scores



```

```python
network = {
    'A': ['B', 'C'],
    'B': ['D'],
    'C': ['B', 'E'],
    'D': ['A'],
    'E': ['D']
}

print("Example Graph:")
for vertex, connections in network.items():
    print(f"{vertex} -> {', '.join(connections)}")
```

```python
pagerank_model = PageRank(network, 10)
pagerank_model.page_ranking()
```
